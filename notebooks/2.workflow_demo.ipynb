# 🧪 Working with Python Files in Jupyter

This notebook demonstrates the recommended workflow for XQubit: write logic in `.py` files under `src/`, and use notebook cells only for running and observing results.

> 🔁 `autoreload` is enabled below — any changes you make to a `.py` file will be picked up automatically without restarting the kernel.

---
## ⚙️ Step 1 — Check Environment

Before anything else, confirm that your `.env` is set up correctly. If you haven't done this yet, go to the project root and run:

```bash
cp .env.example .env
```

Your `.env` should contain:

```dotenv
PYTHONPATH=../src
```

Then **restart the Jupyter server** from inside `pipenv shell` so the `.env` is loaded into the kernel. Run the cell below to verify.

````python
import sys
import os

pythonpath = os.environ.get('PYTHONPATH', 'NOT SET')
src_on_path = any('src' in p for p in sys.path)

print(f"PYTHONPATH     : {pythonpath}")
print(f"src on path    : {src_on_path}")
print(f"Working dir    : {os.getcwd()}")

if not src_on_path:
    print("\n⚠️  src/ is not on your Python path.")
    print("   Make sure your .env contains PYTHONPATH=../src")
    print("   and that you launched Jupyter from inside pipenv shell.")
else:
    print("\n✅ Environment looks good — you're ready to import from src/.")
````

---
## 🔁 Step 2 — Enable Autoreload

This tells Jupyter to watch your `.py` files and reload them automatically whenever you save changes — no kernel restart needed.

````python
%load_ext autoreload
%autoreload 2
````
