## 📖 Introduction of the Study: Quantum Intent Classification

---

### 1. The Linguistic Challenge

In classical NLP, **Intent Classification** works by mapping human language into high-dimensional vector spaces. However, language is inherently **compositional** and **context-dependent** — standard models often struggle to capture complex, non-linear word correlations without massive computational overhead, a bottleneck commonly known as the **"Memory Wall"** phenomenon.

---

### 2. The Quantum Advantage — Hilbert Space Mapping

Quantum Computing offers a novel mathematical framework for text representation. By mapping sentences into **Hilbert Space**, we can treat them as quantum states and exploit two core phenomena:

| Phenomenon | Linguistic Application |
|---|---|
| **Superposition** | A single qubit encodes multiple intent features simultaneously |
| **Entanglement** | Models semantic dependencies between words (e.g., how a modifier shifts verb meaning), inspired by the **DisCoCat** framework |

> DisCoCat — *Distributional Compositional Categorical* — is a model that treats grammatical structure as a tensor network, making it naturally compatible with quantum circuits.

---

### 3. Research Objectives

This study prototypes a **Variational Quantum Classifier (VQC)** to distinguish between user intents (*Greetings* vs. *Farewells*), focusing on two technical challenges:

- **Dimensionality Reduction** — Compressing high-dimensional TF-IDF vectors into a compact **2-qubit** quantum state
- **Noise Resilience** — Evaluating **Error Mitigation** (Resilience Level 1) in a simulated noisy environment modeled after IBM's *Brisbane* quantum architecture
- **Baselinine Benchmarking** — Evaluating the VQC against a classical Support Vector Machine (SVM) to measure the *Quantum-Classical Delta*.

---

### 4. Significance

Successfully classifying intents on a 2-qubit system demonstrates that **Quantum Machine Learning (QML)** can be applied to real-world NLP datasets. Built with **Qiskit 2.x** and Hugging Face, this project serves as a reproducible foundation for future **Quantum-Enhanced Natural Language Understanding (NLU)**.

This study explores the threshold of Principal Component Analysis (PCA) variance required for quantum state preparation, balancing information density with circuit shallow-ness to avoid Barren Plateaus.
By proving that a signal can be extracted from a $2$-dimensional manifold, we provide a blueprint for high-fidelity QNLU on NISQ-era hardware like IBM Brisbane.

````python
%%writefile src/data_loader.py
from datasets import load_dataset
import pandas as pd
from sklearn.model_selection import train_test_split

def fetch_intent_data(labels=[0, 1], samples_per_label=100, test_size=0.2):
    """
    Fetches the Tanaos Intent dataset from Hugging Face.
    Includes auto-detection for column names and version-proof sampling.
    """
    print(f"📡 Requesting dataset: tanaos/synthetic-intent-classifier-dataset-v1...")
    
    # Load the dataset
    raw_dataset = load_dataset("tanaos/synthetic-intent-classifier-dataset-v1", split='train')
    df = raw_dataset.to_pandas()
    
    # Auto-detect and standardize the classification column to 'label'
    if 'intent' in df.columns:
        df = df.rename(columns={'intent': 'label'})
    elif 'labels' in df.columns:
        df = df.rename(columns={'labels': 'label'})
        
    # Failsafe
    if 'label' not in df.columns:
        raise KeyError(f"Critical Error: Could not find classification column. Available: {list(df.columns)}")
        
    # Filter for our target intents (0: greeting, 1: farewell)
    df_filtered = df[df['label'].isin(labels)].copy()
    
    # Version-Proof Class Balancing (Avoids the groupby index trap)
    balanced_frames = []
    for _, group in df_filtered.groupby('label'):
        sampled_group = group.sample(n=min(len(group), samples_per_label), random_state=42)
        balanced_frames.append(sampled_group)
        
    # Reassemble the dataframe safely
    balanced_df = pd.concat(balanced_frames).reset_index(drop=True)
    
    # Split into training and testing sets
    train_df, test_df = train_test_split(
        balanced_df, 
        test_size=test_size, 
        stratify=balanced_df['label'], 
        random_state=42
    )
    
    print(f"✅ Success: {len(train_df)} training and {len(test_df)} testing samples retrieved.")

    print("=" * 100)
    print("  Code for the above program is made by        : John-Ronan S. Beira")
    print("=" * 100)
    return train_df, test_df
````

## 🔡 NLP Feature Engineering: Classical Embedding

To process linguistic data within a quantum architecture, we must first map discrete text strings into a continuous numerical manifold. This stage serves as the **Classical Embedding Layer** of our hybrid pipeline.

### 1. Vectorization via TF-IDF
We utilize **Term Frequency-Inverse Document Frequency (TF-IDF)** to quantify semantic importance. Unlike standard one-hot encoding, TF-IDF effectively filters "noise" words (high-frequency tokens with low semantic value) and highlights discriminative tokens relevant to intent classification. This ensures the quantum circuit receives high-fidelity signal input.

### 2. Dimensionality Reduction & The "Quantum Bottleneck"
Current **Noisy Intermediate-Scale Quantum (NISQ)** devices are constrained by limited qubit counts and gate fidelity. Mapping a 500-dimensional TF-IDF vector directly would require an impractical circuit depth. 
* **The Strategy:** We implement **Principal Component Analysis (PCA)** to project the semantic space onto a 2-dimensional plane ($n=2$). 
* **The Trade-off:** While this reduction entails a loss of variance (information), it ensures the model remains within the coherence limits of our 2-qubit circuit, preventing **Barren Plateaus**—a common failure mode in Quantum Neural Networks where the gradient vanishes during training.

### 3. Data Leakage Prevention
To maintain research integrity, the PCA and TF-IDF models are "fitted" only on the training set. The test set is subsequently "transformed" using these learned parameters. This ensures that the model is evaluated on truly unseen data, accurately simulating a real-world **Quantum-Natural Language Understanding (QNLU)** application.

<div align="right"><i>— John-Ronan S. Beira</i></div>

````python
%%writefile src/nlp_utils.py
import nltk
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import PCA
import pandas as pd

# Ensure NLTK resources are available locally
nltk.download('stopwords', quiet=True)

def process_features(train_df, test_df, n_components=2):
    """
    Transforms raw text into a reduced numerical space for Quantum mapping.
    Uses TF-IDF for vectorization and PCA for dimensionality reduction.
    """
    print(f"🛠️ Vectorizing text and reducing to {n_components} components...")
    
    # 1. Initialize TF-IDF with English stopword removal
    # We limit to 500 features initially to maintain a manageable dense matrix
    stop_words = list(stopwords.words('english'))
    vectorizer = TfidfVectorizer(stop_words=stop_words, max_features=500)
    
    # 2. Fit and transform training data; transform test data using the training vocabulary
    # This prevents 'data leakage' from the test set into our feature space
    X_train_tfidf = vectorizer.fit_transform(train_df['text']).toarray()
    X_test_tfidf = vectorizer.transform(test_df['text']).toarray()
    
    # 3. Principal Component Analysis (PCA)
    # Reducing the semantic space to match our target Qubit count
    pca = PCA(n_components=n_components)
    X_train_pca = pca.fit_transform(X_train_tfidf)
    X_test_pca = pca.transform(X_test_tfidf)
    
    print(f"✅ Feature Engineering Complete.")
    print(f"PCA Variance Explained: {sum(pca.explained_variance_ratio_)*100:.2f}%")
    
    return X_train_pca, X_test_pca, vectorizer, pca

    print("=" * 100)
    print("  Code for the above program is made by        : John-Ronan S. Beira")
    print("=" * 100)
````

````python
%%writefile src/quantum_engine.py
import numpy as np   
from qiskit.circuit.library import ZZFeatureMap, RealAmplitudes
from qiskit_algorithms.optimizers import COBYLA
from qiskit_machine_learning.algorithms import VQC
# Updated imports for Qiskit 1.0+
from qiskit.primitives import StatevectorSampler as Sampler

def build_vqc(num_qubits=2, reps=2, seed=42):
    """
    Constructs the Quantum Circuit components and the VQC object.
    Updated for Qiskit 1.x compatibility.
    """
    # 1. Feature Map: Encodes classical data [x, y] into quantum states
    feature_map = ZZFeatureMap(feature_dimension=num_qubits, reps=reps, entanglement='linear')

    # 2. Ansatz: The trainable 'Quantum Brain' (Hardware-efficient rotations)
    ansatz = RealAmplitudes(num_qubits=num_qubits, reps=reps)

    # 3. Optimizer: Classical algorithm to tune the circuit parameters
    optimizer = COBYLA(maxiter=100)

    # 4. Construct the Variational Quantum Classifier using the new Sampler
    sampler = Sampler()

    num_params = ansatz.num_parameters
    rng = np.random.default_rng(seed)
    initial_point = rng.uniform(-np.pi, np.pi, num_params)

    vqc = VQC(
        sampler=sampler,
        feature_map=feature_map,
        ansatz=ansatz,
        optimizer=optimizer,
        initial_point=initial_point,
    )

    return vqc, feature_map, ansatz

def train_vqc(vqc, X_train, y_train):
    """
    Executes the training loop on the local statevector simulator.
    """
    print("🚀 Initializing Quantum Training Loop...")
    print(f"Training on {len(X_train)} linguistic samples...")

    # vqc.fit expects numpy arrays
    vqc.fit(X_train, y_train)

    print("✅ Quantum Model Training Complete.")
    return vqc

    print("=" * 100)
    print("  Code for the above program is made by        : John-Ronan S. Beira")
    print("=" * 100)
````

## ⚛️ Quantum Architecture & Variational Classifier (VQC)

### 🏗️ Methodological Rationale: Circuit Design
1. **Data Encoding (ZZFeatureMap):** We employ the `ZZFeatureMap` with 2 repetitions. This non-linear encoding method utilizes **Quantum Entanglement** to link our PCA components. In linguistic terms, this captures the "compositional" nature of sentences where the meaning of an intent is derived from the interaction between its components.
2. **The Variational Ansatz (RealAmplitudes):** This serves as the "Quantum Brain." We utilize a hardware-efficient `RealAmplitudes` circuit with 6 trainable parameters. This specific depth is chosen to provide sufficient **Expressivity** for binary classification while keeping the circuit shallow enough to minimize decoherence in a NISQ-era environment.
3. **The Optimization Loop (COBYLA):** For this simulation, we use the **COBYLA** optimizer (100 iterations). As a derivative-free optimizer, it is highly robust against the stochastic "noise" inherent in quantum state sampling, ensuring stable convergence of our 160 training samples.

### 📈 Discussion of Initial Findings: The "Quantum Bias" Analysis

The preliminary evaluation yielded an overall **accuracy of 70%**, which significantly exceeds the baseline of random chance ($50\%$). However, the distribution of errors reveals a nuanced interaction between classical NLP and quantum classification that warrants further optimization.

#### 1. Precision vs. Recall Asymmetry
* **Greeting Precision:** The model showed relatively high precision for **Greetings**, meaning that when it did predict a Greeting, it was more often correct. However, it consistently missed a large portion of actual Greetings, resulting in low recall for this class.
* **Farewell Dominance:** The model consistently captured the majority of **Farewells** across runs, suggesting that Farewell semantics are more separable in the 2-dimensional PCA projection. This likely reflects the 27.76% variance retention, which appears to preserve Farewell-discriminative features better than those of Greetings.

#### 2. The "Defaulting" Phenomenon & The Quantum Bottleneck
The Confusion Matrix highlights a critical area for improvement: **12 Greetings** were misclassified as **Farewells**. 
* **The Information Gap:** By reducing our data to only $27.76\%$ variance, several "Greeting" utterances likely lost their unique linguistic markers, causing them to cluster near the "Farewell" decision boundary. 
* **Class Bias:** In this 2-qubit configuration, the decision hyperplane appears skewed toward the Farewell class. This represents a "Safe-State Bias" where the model defaults to the more semantically distinct category when faced with low-variance input.

#### 3. Transition to Iterative Optimization
